## Here we will extract all the articles content from FRA webpage
URL: https://www.rfa.org/tibetan

```python
def save_json(path, file_name, data):
    """
    
    """
    with open(path+file_name, "w") as outfile:
        json.dump(data, outfile, indent=4)
        # print(f"Successfully saved: {file_name}")
```

```python
def read_json(path, file_name):
    """
    
    """
    with open(path+file_name, 'r') as openfile:
        # Reading from json file
        Loaded_file = json.load(openfile)
        print(f"Successfully loaded: {file_name}")

    return Loaded_file
```

```python
page_key_list = ["གསར་འགྱུར།"]
path_new = "./data/གསར་འགྱུར།/"
file_prefix = "གསར་འགྱུར།"
```

```python
path_old = "./data/"
file_name_old = "RFA_ALL_content_གསར་འགྱུར།.json"
All_links_data = read_json(path_old, file_name_old)
Total_lenght = len(All_links_data)
print(f"Total page in {file_name_old}: {Total_lenght}")
```

```python
index = 0
for each_article_name, each_article_data in tqdm(All_links_data.items()):
    temp_file_name = f"RFA_{file_prefix}_{index}.json"
    save_json(path_new, temp_file_name, each_article_data)
    # print(temp_file_name, each_article_data)
    index += 1
    # break
```
