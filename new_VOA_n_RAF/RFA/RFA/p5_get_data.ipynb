```python
def save_json(path, file_name, data):
    """
    
    """
    with open(path+file_name, "w") as outfile:
        json.dump(data, outfile, indent=4)
        # print(f"Successfully saved: {file_name}")
```

```python
def read_json(path, file_name):
    """
    
    """
    with open(path+file_name, 'r') as openfile:
        # Reading from json file
        Loaded_file = json.load(openfile)
        print(f"Successfully loaded: {file_name}")

    return Loaded_file
```

```python
import os
import json

# Directory where your files are stored
directory = path_new  # Use the same path variable from your original code

# Get all files in the directory
files = os.listdir(directory)

# Filter for files that should be JSON but don't have the extension
for file in files:
    # Check if this is one of your data files without .json extension
    if file.startswith("RFA_" + file_prefix) and not file.endswith('.json'):
        # Full path to the original file
        original_file_path = os.path.join(directory, file)
        
        # New file path with .json extension
        new_file_path = original_file_path + '.json'
        
        try:
            # Read the file content
            with open(original_file_path, 'r', encoding='utf-8') as f:
                content = f.read()
            
            # Parse it as JSON
            data = json.loads(content)
            
            # Save it back with the correct extension
            with open(new_file_path, 'w', encoding='utf-8') as f:
                json.dump(data, f, ensure_ascii=False, indent=4)
            
            print(f"Successfully converted {file} to {file}.json")
            
            # Optionally remove the original file
            # os.remove(original_file_path)
            
        except json.JSONDecodeError:
            print(f"Error: {file} does not contain valid JSON data")
        except Exception as e:
            print(f"Error processing {file}: {str(e)}")
```
