```python
import pandas as pd
import json 
import os
```

```python

def extract_audio_links(folder_path, output_file):
    """
    Extract audio links from JSON files in the specified folder
    and create a combined JSON with {filename: audio_link} format.
    """
    # Dictionary to store filename: audio link pairs
    audio_links = {}
    
    # Check if the folder exists
    if not os.path.exists(folder_path):
        print(f"Error: Folder '{folder_path}' not found.")
        return None
    
    # Iterate through all files in the folder
    for filename in os.listdir(folder_path):
        # Only process JSON files
        if filename.endswith('.json'):
            file_path = os.path.join(folder_path, filename)
            
            try:
                # Read and parse the JSON file
                with open(file_path, 'r', encoding='utf-8') as file:
                    try:
                        data = json.load(file)
                        
                        # Extract the audio link
                        if (
                            'data' in data and 
                            'body' in data['data'] and 
                            'URL' in data['data']['meta_data']
                        ):
                            audio_link = data['data']['meta_data']['URL']
                            
                            # Only add to the dictionary if there's a valid audio link
                            if audio_link and audio_link != "":
                                # Remove the .json extension from the filename
                                filename_without_extension = os.path.splitext(filename)[0]
                                audio_links[filename_without_extension] = audio_link
                                
                    except json.JSONDecodeError:
                        print(f"Error: Unable to parse JSON in file '{filename}'")
            except Exception as e:
                print(f"Error processing file '{filename}': {str(e)}")
    
    # Write the combined data to a new JSON file
    
    # output_file = f"{categ}_combined_audio_links.json"
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(audio_links, f, ensure_ascii=False, indent=4)
    
    print(f"Successfully created '{output_file}' with {len(audio_links)} all article links.")
    return audio_links, len(audio_links)
```

```python
categ = "བོད།"
folder_path = f"./new_data/{categ}/"
file_name = f"{categ}_combined_all_articles_links.json"
output_file = "./new_data/" + file_name

first_links, file_with_audio = extract_audio_links(folder_path, output_file)
```

```python
import pandas as pd
import shutil
import os


# Define the clean data destination directory
clean_data_dir = './new_data/clean_data/'

# Create the clean data directory if it doesn't exist
os.makedirs(clean_data_dir, exist_ok=True)

# Iterate through the DataFrame rows
for index, row in clean_merged_df.iterrows():
    # Construct source and destination file paths
    json_source = f"{row['data_path']}{row['file_name']}.json"
    audio_source = f"{row['audio_path']}{row['file_name']}.mp3"
    
    json_dest = f"{clean_data_dir}{row['file_name']}.json"
    audio_dest = f"{clean_data_dir}{row['file_name']}.mp3"
    
    # Copy JSON file
    try:
        shutil.copy2(json_source, json_dest)
        print(f"Copied JSON: {json_source} -> {json_dest}")
    except FileNotFoundError:
        print(f"JSON file not found: {json_source}")
    except Exception as e:
        print(f"Error copying JSON {json_source}: {str(e)}")
    
    # Copy MP3 file
    try:
        shutil.copy2(audio_source, audio_dest)
        print(f"Copied MP3: {audio_source} -> {audio_dest}")
    except FileNotFoundError:
        print(f"MP3 file not found: {audio_source}")
    except Exception as e:
        print(f"Error copying MP3 {audio_source}: {str(e)}")

print("File copying complete!")
```
